La pandilla de las manzanas: Imagina que un grupo de amigos quiere saber mas sobre su cosecha de Manzanas. Cosechan 10 manzanas y quieren analizar su peso, para ver como es su producción de este año.
Queremos calcular:
- El peso promedio de las manzanas(Media)
- El peso central (Mediana)
- El peso mas común (Moda)
Desviación estandar = Qué tanto varian los pesos

```python
import numpy as np
import statistics as stats

#Pesos de las manzanas en gramos
manzanas = [150, 180, 200, 170, 150, 190, 200, 210, 150, 180]

#Estadísticas descriptivas
media = np.mean(manzanas) # promedio
mediana = np.median(manzanas) #valor central
moda = stats.mode(manzanas) #valor mas común
desviacion = np.std(manzanas) #Variabilidad de los pesos

#Resultados
print(f"Analisis de la cosecha de manzanas")
print(f"Peso promedio: {media} gramos")
print(f"Peso central: {mediana} gramos")
print(f"Peso mas común: {moda} gramos")
print(f"Desviación estandar: {desviacion:.2f} gramos")
```
